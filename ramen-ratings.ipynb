```python
import pandas as pd

df = pd.read_csv('ramen-ratings.csv')

print(df.head())
print(df.shape)
print(df.columns)
print(df.info())
print(df.describe())
print(df.isnull().sum())
```

```python
# What are the different ramen styles?
print(df['Style'].value_counts())

# Which countries have the most ramen brands?
print(df['Country'].value_counts().head(10))

# What is the average rating?
# Convert Stars column to numeric, errors='coerce' turns invalid values into NaN
df['Stars'] = pd.to_numeric(df['Stars'], errors='coerce')

# Now this will work
print(df['Stars'].mean())

# What is the highest rated ramen?
print(df[df['Stars'] == df['Stars'].max()])
```

```python
# Check for missing values
print(df.isnull().sum())

# Drop rows with missing values
df = df.dropna()
```

```python
import matplotlib.pyplot as plt

# Top 10 countries by number of ramen brands
df['Country'].value_counts().head(10).plot(kind='bar', color='orange')
plt.title('Top 10 Countries by Ramen Brands')
plt.xlabel('Country')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```
